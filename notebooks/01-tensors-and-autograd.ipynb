```python
import torch

print("Pytorch", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("VRAM:", torch.cuda.get_device_properties(0).total_memory / 1024**3, "GB")
```

```python
a = torch.tensor(
    [[1.,2.],
    [3., 4.]],
    device="cuda"
)

b = torch.tensor(
    [[5., 6.],
    [7., 8.]],
    device="cuda"
)

c = a @ b

print(c)
print(c.device)

x = torch.tensor(3.0, requires_grad=True)

y = x**2 + 2*x + 1

y.backward()

print("y:", y.item())
print("dy/dx:", x.grad.item())
```

What is a tensor? 
- A tensor is a object in the torch module that stores values as a scalar (1 number), a vector (1 row), a matrix (2d), or multidimensional array. 

What is the difference between a scalar, vector, and matrix?
- A scalar is a single number. A vector is a single row of numbers. A matrix is a 2 dimensional array of numbers.

What does a @ b mean?
- This annotation means do matrix multiplication on a & b. There is a major difference between this and doing a * b.

What does requires_grad=True mean? 
- This tells python to keep track of the operations performed on this tensor, in other words how much the variable is affecting the output on each pass so the gradient can be calculated later.

What does y.backward() actually calculate? 
- This method looks at where the variable has arrived at and does backwards operations until the starting point. It calculates how the gradient can effect the outcome.
