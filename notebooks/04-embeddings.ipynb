## Goals
- Understand what an embedding is
- Understand why AI represents things as vectors
- Understand token IDs vs embedding vectors
- Understand `nn.Embedding`
- Understand embedding dimensions
- Understand that embedding values are learned parameters
- Understand vector similarity
- Understand cosine similarity conceptually
- Understand why similar meanings can end up near each other in embedding space

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# ------------
# CREATE A SMALL VOCABULARY
# --------------------------

words = [
    "dog",
    "cat",
    "truck",
    "car",
    "apple",
]

word_to_id = {
    "dog": 0,
    "cat": 1,
    "truck": 2,
    "car": 3,
    "apple": 4
}

# -----------
# CREATE EMBEDDING LAYER
# ----------------------

embedding = nn.Embedding(
    num_embeddings=5,
    embedding_dim=3
)

# --------------------------
# LOOK UP ONE WORD
# ------------------------

dog_id = torch.tensor(word_to_id["dog"])

dog_embedding = embedding(dog_id)

print("Dog ID: ")
print(dog_id)

print("\nDog embedding: ")
print(dog_embedding)

#----------------
# LOOK UP MULTIPLE WORDS
# -----------------------

word_ids = torch.tensor([
    word_to_id["dog"],
    word_to_id["cat"],
    word_to_id["truck"],

])

vectors = embedding(word_ids)

print("\nMultiple embeddings:")
print(vectors)

print("\nShape:")
print(vectors.shape)

# -------------------
# INSPECT ALL EMBEDDINGS
# ---------------------

print("\nEmbedding table:")
print(embedding.weight)

# ---------------------
# COMPARE TWO VECTORS
# -------------------

dog_vector = embedding(
    torch.tensor(word_to_id["dog"])
)

cat_vector = embedding(
    torch.tensor(word_to_id["cat"])
)

similarity = F.cosine_similarity(
    dog_vector.unsqueeze(0),
    cat_vector.unsqueeze(0)
)

print("\nDog / cat cosine similarity:")
print(similarity.item())
```
